```python

import pandas as pd

customers = pd.read_csv(
    r"D:\Customer_Segmentation_BI\data\processed\customer_segments.csv"
)

segment_names = {
    0: (
        "Regular-value customers",
        "Customers with moderate purchasing activity",
    ),
    1: (
        "High-value loyal customers",
        "Frequent purchasers with high spending",
    ),
    2: (
        "At-risk / inactive customers",
        "Customers who have not purchased recently",
    ),
}

profiles = (
    customers.groupby("Cluster")
    .agg(
        customer_count=("CustomerID", "count"),
        total_revenue=("Monetary", "sum"),
        avg_recency=("Recency", "mean"),
        avg_frequency=("Frequency", "mean"),
        avg_monetary=("Monetary", "mean"),
    )
    .reset_index()
)

total_revenue = customers["Monetary"].sum()

profiles["segment_name"] = profiles["Cluster"].map(
    lambda c: segment_names[int(c)][0]
)
profiles["description"] = profiles["Cluster"].map(
    lambda c: segment_names[int(c)][1]
)
profiles["revenue_percentage"] = (
    profiles["total_revenue"] / total_revenue * 100
).round(2)

profiles = profiles.rename(columns={"Cluster": "cluster"})

columns = [
    "cluster",
    "segment_name",
    "description",
    "customer_count",
    "total_revenue",
    "revenue_percentage",
    "avg_recency",
    "avg_frequency",
    "avg_monetary",
]

output_path = (
    r"D:\Customer_Segmentation_BI\data\processed\cluster_profiles_import.csv"
)

profiles[columns].to_csv(output_path, index=False)

print(profiles[columns])
print("\nSaved to:", output_path)
```
